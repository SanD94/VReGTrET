```r
# Define a custom transformation (example: square root)
sqrt_trans <- new_transform(
  name = "sqrt",
  transform = function(x) 2 * x,
  inverse = function(x) x,
  breaks = function(x) seq(0, 35, by = 5)
)

# Apply it to scale_x_continuous
ggplot(mtcars, aes(mpg, wt)) +
  geom_point() +
  scale_x_continuous(trans = sqrt_trans, limits = c(0, 35))
```
